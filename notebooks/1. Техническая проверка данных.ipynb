##### 1.2.1) Проверка для идентификаторов: 'item_id', 'increment_id', 'sku', 'Customer ID', 'sku'

Некоторые столбцы содержат в себе значение "\N", которое означает отсутствующие данные. Такие значения заменим на pd.NA.

```python
id_cols = [
    "item_id",
    "increment_id",
    "Customer ID",
    "sku",
    "sales_commission_code"
]

for col in id_cols:
    s = df[col]
    s = s.str.strip()
    s = s.replace(r"\N", pd.NA)

    print(f"\n--- {col} ---")
    print("Пропусков:", s.isna().sum())
    df[col] = df[col].astype("string").str.strip()
    print("Уникальных значений:", s.nunique(dropna=True))
    print("Повторяющихся значений:",
          s[s.notna()].duplicated().sum())

    df[col] = s
```

##### 1.2.2) Проверка для числовых данных: 'price', 'qty_ordered', 'grand_total', 'discount_amount', 'MV'

```python
numeric_cols = [
    "price",
    "qty_ordered",
    "grand_total",
    "discount_amount",
    "MV"
]

for col in numeric_cols:
    numeric = pd.to_numeric(df[col], errors="coerce")

    bad_mask = df[col].notna() & numeric.isna()

    print(f"\n--- {col} ---")
    print("Некорректных:", bad_mask.sum())
    print("Пропусков:", df[col].isna().sum())
    print("Нулевых:", (numeric == 0).sum())
    print("Отрицательных:", (numeric < 0).sum())
```

```python
mv_num = pd.to_numeric(df["MV"], errors="coerce")
bad_mask = df["MV"].notna() & mv_num.isna()
df.loc[bad_mask, "MV"].value_counts().head(20)
```

Заметим, что здесь используется разделитель в виде запятой. Изначально я предположила, что он используется как разделитель дробной и целой части, но в данных присутствовали значения вида "1,000,000", поэтому запятая используется как разделитель тысяч. Также заметим, что здесь присутствуют пропуски в данных, отмеченные символом "-".

```python
mv_clean = (
    df["MV"]
    .astype("string")
    .str.strip()
    .str.replace(",", "", regex=False)
)

mv_clean = mv_clean.replace("-", pd.NA)

mv_num = pd.to_numeric(mv_clean, errors="coerce")
bad_mask = mv_clean.notna() & mv_num.isna()

print("Некорректных после очистки:", bad_mask.sum())
print(
    mv_clean[bad_mask]
    .value_counts()
    .head(20)
)
```

Проверка подходит, поэтому можем выполнить её на целом датасете.

```python
df["MV"] = (
    df["MV"]
    .astype("string")
    .str.strip()
    .str.replace(",", "", regex=False)
    .replace("-", pd.NA)
)

df["MV"] = pd.to_numeric(df["MV"], errors="coerce")
```

```python
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

##### 1.2.3) Проверка столбцов с категориальными данными: 'status', 'category_name_1', 'payment_method', 'BI Status'

```python
categorical_cols = [
    "status",
    "category_name_1",
    "payment_method",
    "BI Status"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

Заметим, что здесь также встречается "\N", а также "#REF!" - это, скорее всего, значение из Excel, содержащее некорректную ссылку. Заменим это значение на отсутствующее.

```python
df["status"] = df["status"].replace(r"\N", pd.NA)
df["category_name_1"] = df["category_name_1"].replace(r"\N", pd.NA)
df["BI Status"] = df["BI Status"].replace("#REF!", pd.NA)
```

##### 1.2.4) Проверка столбцов для записи дат: 'created_at', 'Working Date', 'Customer Since'

```python
date_cols = [
    "created_at",
    "Working Date",
    "Customer Since"
]

for col in date_cols:
    parsed = pd.to_datetime(df[col], errors="coerce")

    bad_mask = df[col].notna() & parsed.isna()

    print(f"\n--- {col} ---")
    print("Некорректных дат:", bad_mask.sum())
    print("Пропусков:", df[col].isna().sum())
    print("Минимальная дата:", parsed.min())
    print("Максимальная дата:", parsed.max())
```

```python
df["created_at"] = pd.to_datetime(
    df["created_at"],
    errors="coerce"
)
df["Working Date"] = pd.to_datetime(
    df["Working Date"],
    errors="coerce"
)

df["Customer Since"] = pd.to_datetime(
    df["Customer Since"],
    errors="coerce"
)
```

##### 1.2.5) Проверка производных от даты столбцов: 'Year', 'Month', 'M-Y', 'FY'

```python
derived_date_cols = [
    "Year",
    "Month",
    "M-Y",
    "FY"
]

for col in derived_date_cols:
    print(f"\n--- {col} ---")
    print("Пропусков:", df[col].isna().sum())
    print("Уникальных:", df[col].nunique(dropna=True))
    print(df[col].value_counts(dropna=False).head(20))
```

Проверим, что они совпадают с Working Date.

```python
year_num = pd.to_numeric(df["Year"], errors="coerce")

mask_year_mismatch = (
    df["Working Date"].notna()
    & year_num.notna()
    & (year_num != df["Working Date"].dt.year)
)

print("Несоответствий Year:", mask_year_mismatch.sum())
```

```python
month_num = pd.to_numeric(df["Month"], errors="coerce")

mask_month_mismatch = (
    df["Working Date"].notna()
    & month_num.notna()
    & (month_num != df["Working Date"].dt.month)
)

print("Несоответствий Month:", mask_month_mismatch.sum())
```

```python
expected_my = (
    df["Working Date"].dt.month.astype("Int64").astype("string")
    + "-"
    + df["Working Date"].dt.year.astype("Int64").astype("string")
)

mask_my_mismatch = (
    df["Working Date"].notna()
    & df["M-Y"].notna()
    & (df["M-Y"].astype("string") != expected_my)
)

print("Несоответствий M-Y:", mask_my_mismatch.sum())
```

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")
df["Month"] = pd.to_numeric(df["Month"], errors="coerce").astype("Int64")
```
